```python
from typing import Literal
from pydantic import BaseModel, ConfigDict, Field


class Ticket(BaseModel):
    model_config = ConfigDict(extra="forbid")

    customer_name: str | None = None
    order_id: str | None = None
    product: str | None = None
    amount: float | None = Field(default=None, ge=0)

    issue_type: Literal[
        "Damaged Product",
        "Defective Product",
        "Delayed Delivery",
        "Billing Issue",
        "Other",
    ]

    requested_action: Literal[
        "Refund",
        "Replacement",
        "Repair",
        "Information",
        "Other",
    ]

    priority: Literal["Low", "Normal", "High"]


sample = {
    "customer_name": "Rahul",
    "order_id": "ORD-1234",
    "product": None,
    "amount": 2499,
    "issue_type": "Damaged Product",
    "requested_action": "Replacement",
    "priority": "Normal",
}

ticket = Ticket.model_validate(sample)
print("Valid ticket!")
print(ticket.model_dump())
```

```python
import re

def extract_ticket(text):
    message = text.lower()

    # Extract order ID
    order_match = re.search(r"\bORD-\d+\b", text, re.I)
    order_id = order_match.group(0).upper() if order_match else None

    # Extract amount when explicitly stated
    amount_match = re.search(
        r"(?:paid|amount|cost|price)\s*(?:of|was|is|:)?\s*₹?\s*(\d+(?:\.\d{1,2})?)",
        text,
        re.I,
    )
    amount = float(amount_match.group(1)) if amount_match else None

    # Extract customer name from "I'm NAME" or "I am NAME"
    name_match = re.search(
        r"\b(?:I'm|I am)\s+([A-Z][a-z]+)\b", text
    )
    customer_name = name_match.group(1) if name_match else None

    # Classify issue
    if "damaged" in message:
        issue_type = "Damaged Product"
    elif any(word in message for word in ["stopped working", "defective", "not working"]):
        issue_type = "Defective Product"
    elif any(word in message for word in ["late", "delayed"]):
        issue_type = "Delayed Delivery"
    elif any(word in message for word in ["billing", "charged twice", "wrong charge"]):
        issue_type = "Billing Issue"
    else:
        issue_type = "Other"

    # Classify requested action
    if any(word in message for word in ["refund", "money back"]):
        requested_action = "Refund"
    elif any(word in message for word in ["replacement", "replace"]):
        requested_action = "Replacement"
    elif "repair" in message:
        requested_action = "Repair"
    elif any(word in message for word in ["information", "tell me", "explain"]):
        requested_action = "Information"
    else:
        requested_action = "Other"

    # Simple baseline priority rule
    priority = "High" if any(
        word in message for word in ["urgent", "emergency", "immediately"]
    ) else "Normal"

    # Simple product extraction for this starter demo
    product = "headphones" if "headphones" in message.lower() else None

    result = {
        "customer_name": customer_name,
        "order_id": order_id,
        "product": product,
        "amount": amount,
        "issue_type": issue_type,
        "requested_action": requested_action,
        "priority": priority,
    }

    # Validate output against our schema
    return Ticket.model_validate(result).model_dump()


message = (
    "Hi, I'm Rahul. Order ORD-1234 arrived damaged. "
    "I paid 2499 and want a replacement."
)

import json
print(json.dumps(extract_ticket(message), indent=2))
```

```python
%cd /content/Syntax-error-ps5

!cp app/main.py app/main.py.bak

%%writefile /content/Syntax-error-ps5/app/main.py
import json
import re

import torch
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, ValidationError
from app.schema import Ticket

app = FastAPI(title="TicketStruct AI")

# These are loaded once when the API module starts.
MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    dtype=torch.float16,
    device_map="auto",
)

class ExtractRequest(BaseModel):
    text: str

@app.get("/")
def home():
    return {"project": "TicketStruct AI", "status": "running"}

@app.post("/validate")
def validate_output(payload: dict):
    try:
        result = Ticket.model_validate(payload)
        return {"valid": True, "data": result.model_dump()}
    except ValidationError as exc:
        raise HTTPException(status_code=422, detail=exc.errors())

@app.post("/extract")
def extract_ticket(request: ExtractRequest):
    message = request.text

    prompt = f"""Extract a customer support ticket from the message.
Return only one JSON object with these exact fields:
customer_name, order_id, product, amount, issue_type, requested_action, priority.

Use null for missing values; never use "unknown".
issue_type must be one of: Damaged Product, Defective Product,
Delayed Delivery, Billing Issue, Other.
requested_action must be one of: Refund, Replacement, Repair, Information, Other.
priority must be Low, Normal, or High.
Set priority to High only when the message explicitly indicates urgency,
such as urgent, emergency, or immediately. Otherwise use Normal.
Do not infer a product that is not stated.

Customer message: {message}
JSON:
"""

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=180,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id,
        )

    generated = output[0][inputs["input_ids"].shape[1]:]
    answer = tokenizer.decode(generated, skip_special_tokens=True)

    match = re.search(r"\{.*?\}", answer, re.DOTALL)
    if not match:
        raise HTTPException(status_code=502, detail="Model did not return JSON")

    try:
        data = json.loads(match.group(0))
        if isinstance(data.get("product"), str) and any(
            phrase in data["product"].lower()
            for phrase in ("unknown", "not specified", "no product mentioned")
        ):
            data["product"] = None

        # Apply the explicit baseline urgency rule consistently.
        urgent_words = ("urgent", "emergency", "immediately")
        data["priority"] = (
            "High" if any(word in message.lower() for word in urgent_words)
            else "Normal"
        )

        ticket = Ticket.model_validate(data)
        return {"valid": True, "data": ticket.model_dump()}
    except (json.JSONDecodeError, ValidationError) as exc:
        raise HTTPException(status_code=422, detail=str(exc))
```
